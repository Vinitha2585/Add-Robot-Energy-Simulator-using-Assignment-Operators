```python
#Robot Energy stimulator using assignment operator 

energy=int(input("Enter the energy:"))
walking=int(input("Enter the walking :"))
lifting=int(input("Enter the lifting:"))
charging=int(input("Enter the charging:"))
solar_mode=int(input("Enter the solar mode:"))
task_division=int(input("Enter the task division:"))
backup_energy=int(input("Enter the backup energy:"))

energy-=walking #subraction and assign
print("waking energy:",energy)

energy-=lifting #subraction and assign
print("lifting energy:",energy)

energy+=charging #Addition and assign
print("charging:",energy)

energy*=solar_mode #Multiplication assign
print("solar mode:",energy)

energy//=task_division #floor division and assign
print("task division:",energy)

energy%=backup_energy #modulus and assign
print("backup energy:",energy)
```
